```python
import pandas as pd
import numpy as np

cols = [
    "Unnamed: 0", "cat2_slug", "cat3_slug", "city_slug",
    "has_business_deed", "has_elevator", "has_warehouse", "has_parking",
    "has_water", "has_electricity", "has_gas", "has_security_guard",
    "has_barbecue", "has_pool", "has_jacuzzi", "has_sauna", "is_rebuilt",
    "has_balcony", "has_warm_water_provider", "has_heating_system",
    "has_cooling_system", "has_restroom", "rent_to_single",
    "created_at_month",
    "location_latitude", "location_longitude", "location_radius",
]


df = pd.read_csv("../data/Divar.csv", usecols=cols, dtype=str)

df = df.rename(columns={"Unnamed: 0": "row_id"})
df["row_id"] = df["row_id"].astype(int)

print(df.shape)
df.head()
```

```python
bool_cols = [
    "has_business_deed", "has_elevator", "has_warehouse", "has_parking",
    "has_water", "has_electricity", "has_gas", "has_security_guard",
    "has_barbecue", "has_pool", "has_jacuzzi", "has_sauna", "is_rebuilt",
    "has_balcony",
]

for col in bool_cols:
    print(col)
    print(df[col].value_counts(dropna=False))
    print()
```

```python
for col in bool_cols:
    df[col] = df[col].str.lower()
    df[col] = df[col].replace("unselect", np.nan)
    df[col] = df[col].map({"true": True, "false": False})
    df[col] = df[col].astype("boolean")

for col in bool_cols:
    print(col)
    print(df[col].value_counts(dropna=False))
    print()
```

```python
multi_cols = ["has_warm_water_provider", "has_heating_system",
              "has_cooling_system", "has_restroom"]

df[multi_cols] = df[multi_cols].replace("unselect", np.nan)

for col in multi_cols:
    counts = df[col].value_counts(dropna=False)
    percent = (counts / len(df) * 100).round(1)
    display(pd.DataFrame({"count": counts, "%": percent}).rename_axis(col))
```

```python
fill = (
    df[bool_cols].notna()
    .groupby(df["cat3_slug"])
    .mean()
    .mul(100)
    .round()
    .astype(int)
)

fill
```

```python
import jdatetime
df["created_at_month"] = pd.to_datetime(df["created_at_month"])

month_names = ["فروردین", "اردیبهشت", "خرداد", "تیر", "مرداد", "شهریور",
               "مهر", "آبان", "آذر", "دی", "بهمن", "اسفند"]

to_jalali = {}
for m in df["created_at_month"].unique():
    mid_month = (m + pd.Timedelta(days=14)).date()
    to_jalali[m] = jdatetime.date.fromgregorian(date=mid_month)

jalali = df["created_at_month"].map(to_jalali)

df["jalali_year"] = jalali.map(lambda d: d.year)
df["jalali_month"] = jalali.map(lambda d: d.month)
df["jalali_label"] = (df["jalali_year"].astype(str) + " "
                      + df["jalali_month"].map(lambda m: month_names[m - 1]))

df.groupby(["created_at_month", "jalali_label"]).size().to_frame("ads")
```

```python
df["is_main_period"] = df["created_at_month"].between("2024-05-01", "2024-12-01")

first = df["created_at_month"].min()
df["month_index"] = ((df["created_at_month"].dt.year - first.year) * 12
                     + (df["created_at_month"].dt.month - first.month))

share = df["is_main_period"].mean() * 100
print(f"سهم دوره‌ی اصلی: {share:.1f}%")

df[["created_at_month", "jalali_label", "month_index", "is_main_period"]].drop_duplicates().sort_values("month_index").tail(10)
```

```python
lat, lon = "location_latitude", "location_longitude"
df[lat] = df[lat].astype(float)
df[lon] = df[lon].astype(float)

df["coord_status"] = "ok"
df.loc[df[lat].isna(), "coord_status"] = "missing"

inside_iran = df[lat].between(25, 40) & df[lon].between(44, 64)
outside = df[lat].notna() & ~inside_iran

df.loc[outside, "coord_status"] = "kharej"
df.loc[outside, [lat, lon]] = np.nan

df["coord_status"].value_counts()
```

```python
has_coord = df[lat].notna()
pins = df[has_coord].groupby([lat, lon])

df["pin_repeat_count"] = pins[lat].transform("size").astype("Int64")
cities_per_pin = pins["city_slug"].transform("nunique")

bad_idx = cities_per_pin[cities_per_pin > 1].index
df.loc[bad_idx, "coord_status"] = "pin chand shahri"
df.loc[bad_idx, [lat, lon]] = np.nan

display(df["coord_status"].value_counts())
display(pins.size().sort_values(ascending=False).head(10).to_frame("ads"))
```

```python
def haversine_km(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    a = np.sin((lat2 - lat1) / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2) ** 2
    return 6371 * 2 * np.arcsin(np.sqrt(a))
city_counts = df.groupby("city_slug")[lat].count()
centers = df.groupby("city_slug")[[lat, lon]].median()
centers = centers[city_counts >= 30]

center_lat = df["city_slug"].map(centers[lat])
center_lon = df["city_slug"].map(centers[lon])

df["dist_to_city_km"] = haversine_km(df[lat], df[lon], center_lat, center_lon).round(1)

far = df["dist_to_city_km"] > 50
df.loc[far, "coord_status"] = "door_az_shahr"
df.loc[far, [lat, lon]] = np.nan

display(df["coord_status"].value_counts())
display(df[far].groupby("city_slug").size().sort_values(ascending=False).head(10).to_frame("ads"))
```

```python
out_path = "../data/pre_processing_shahab.parquet"
df = df.drop(columns=["rent_to_single", "location_radius"], errors="ignore")

out_cols = (["row_id"] + bool_cols + multi_cols
            + ["created_at_month", "jalali_year", "jalali_month", "jalali_label",
               "month_index", "is_main_period",
               lat, lon, "coord_status", "dist_to_city_km", "pin_repeat_count"])

df[out_cols].to_parquet(out_path, index=False)

check = pd.read_parquet(out_path)
print(check.shape)
check.dtypes
```
