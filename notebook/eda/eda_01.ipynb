```python
import numpy as np 
import pandas as pandas
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
#load_data 
df = pd.read_parquet('../../data/Divar_cleaned.parquet')
```

```python
cat2_counts = df['cat2_slug'].value_counts()
print(cat2_counts)
```

```python
cat3_counts = df['cat3_slug'].value_counts()
print(cat3_counts)
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

cat2_counts = df["cat2_slug"].value_counts()

sns.barplot(x=cat2_counts.values, y=cat2_counts.index, ax=ax, palette="viridis")

ax.bar_label(ax.containers[0], padding=5)

ax.set_title("Distribution of Ads by Category Level 2")
ax.set_xlabel("Number of Ads")
ax.set_ylabel("Category")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

cat3_counts = df["cat3_slug"].value_counts()

sns.barplot(x=cat3_counts.values, y=cat3_counts.index, ax=ax, palette="mako")

ax.bar_label(ax.containers[0], padding=5)

ax.set_title("Distribution of Ads by Category Level 3")
ax.set_xlabel("Number of Ads")
ax.set_ylabel("Category")

plt.tight_layout()
plt.show()
```
