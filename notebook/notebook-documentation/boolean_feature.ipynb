```python
#has_balcony

df['has_balcony'] = df['has_balcony'].astype(str).str.lower().str.strip()

df['has_balcony'] = df['has_balcony'].replace({
    'true': True,
    'false': False,
    'unselect': np.nan,
    'nan': np.nan
})

print(df['has_balcony'].value_counts(dropna=False))

df['has_balcony'].isnull().sum() / len(df['has_balcony']) * 100
```

```python
# drop feature hase more than 95% Null value 

drop_bool_cols = [
    'has_business_deed', 'has_water', 'has_electricity', 'has_gas',
    'has_security_guard', 'has_barbecue', 'has_pool', 'has_jacuzzi', 'has_sauna'
]

df = df.drop(columns=drop_bool_cols, errors='ignore')
```

```python
other_main_bools = ['has_elevator', 'has_warehouse', 'has_parking', 'is_rebuilt']

for col in other_main_bools:
    df[col] = df[col].astype(str).str.lower().str.strip().map({
        'true': True,
        'false': False,
        'nan': np.nan,
        'none': np.nan
    })
```

```python
bool_summary = pd.DataFrame({
    'Null_Count': df[other_main_bools].isnull().sum(),
    'Null_Percentage': (df[other_main_bools].isnull().mean() * 100).round(2) ,
    'True_Count': (df[other_main_bools] == True).sum(),
    'False_Count': (df[other_main_bools] == False).sum()
})

print(bool_summary)
```
