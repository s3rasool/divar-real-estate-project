### آزمون فرض: مقایسه مساحت خانه‌های قدیمی‌ساخت و جدیدساخت

با توجه به این تصور رایج که «خانه‌های قدیمی دلبازتر بودند»، بررسی می‌کنیم که آیا میانگین مساحت خانه‌های قدیمی‌ساخت بیشتر از خانه‌های جدیدساخت است یا خیر.

در این تحلیل:

- خانه‌های **قدیمی‌ساخت**: خانه‌هایی که قبل از سال ۱۳۹۶ ساخته شده‌اند.
- خانه‌های **جدیدساخت**: خانه‌هایی که در سال ۱۳۹۶ یا بعد از آن ساخته شده‌اند.
- متغیر مورد مقایسه: `building_size`

فرضیه‌های آماری به صورت زیر تعریف می‌شوند:

**فرض صفر (H₀):** میانگین مساحت خانه‌های قدیمی‌ساخت بیشتر از خانه‌های جدیدساخت نیست.

**فرض مقابل (H₁):** میانگین مساحت خانه‌های قدیمی‌ساخت بیشتر از خانه‌های جدیدساخت است.

</div>

```python
import pandas as pd
import numpy as np 
```

```python
df = pd.read_pickle("../../data/Divar_cleaned.pkl")
```

```python
housing_df = df[
    df["cat2_slug"].isin([
        "residential-sell",
        "residential-rent"
    ])
    & df["construction_year"].notna()
    & df["building_size"].notna()
].copy()

print("Rows:", len(housing_df))

print(
    housing_df[
        ["cat2_slug", "construction_year", "building_size"]
    ].head()
)
```

```python
old_houses = housing_df[
    housing_df["construction_year"] < 1396
].copy()

new_houses = housing_df[
    housing_df["construction_year"] >= 1396
].copy()

print("Old houses:", len(old_houses))
print("New houses:", len(new_houses))

print("\nOld mean building size:", old_houses["building_size"].mean())
print("New mean building size:", new_houses["building_size"].mean())

print("\nOld median building size:", old_houses["building_size"].median())
print("New median building size:", new_houses["building_size"].median())
```

```python
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(
    old_houses["building_size"],
    new_houses["building_size"],
    equal_var=False,
    alternative="greater"
)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

### نتیجه‌گیری آزمون فرض

برای بررسی این ادعا که «خانه‌های قدیمی‌ساخت به‌طور میانگین مساحت بیشتری نسبت به خانه‌های جدیدساخت دارند»، خانه‌ها به دو گروه تقسیم شدند:

- قدیمی‌ساخت: سال ساخت کمتر از ۱۳۹۶
- جدیدساخت: سال ساخت ۱۳۹۶ و بعد از آن

میانگین مساحت خانه‌های قدیمی‌ساخت برابر با حدود **۱۰۶.۶ مترمربع** و میانگین مساحت خانه‌های جدیدساخت برابر با حدود **۱۲۶.۰ مترمربع** به‌دست آمد.

برای مقایسه میانگین دو گروه از آزمون **Welch's t-test یک‌طرفه** استفاده شد.

نتایج آزمون:

- `t-statistic = -106.00`
- `p-value = 1.0`

با توجه به اینکه مقدار `p-value` بسیار بزرگ‌تر از سطح معناداری ۰.۰۵ است، فرض صفر رد نمی‌شود.

بنابراین، داده‌های این مجموعه شواهدی برای تأیید این فرضیه که خانه‌های قدیمی‌ساخت به‌طور میانگین بزرگ‌تر از خانه‌های جدیدساخت هستند، ارائه نمی‌کنند.

در واقع، در این داده‌ها میانگین مساحت خانه‌های جدیدساخت بیشتر از خانه‌های قدیمی‌ساخت مشاهده شده است.
