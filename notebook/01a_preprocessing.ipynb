```python
import numpy as np 
import pandas as pd 

df = pd.read_csv("/home/rasool-estiri/artificial-intelligence/project/real-estate/data/Divar.csv")
city_classification = pd.read_csv("/home/rasool-estiri/artificial-intelligence/project/real-estate/data/iran_city_classification.csv")

print(df.columns.tolist)
```

```python
# Category, Location & Information
cols = ['cat2_slug', 'cat3_slug', 'city_slug', 'neighborhood_slug', 
        'user_type', 'deed_type', 
        'building_direction', 'floor_material']

for col in cols : 
    df[col] = (df[col].astype(str)
                      .str.strip()
                      .str.replace('ي', 'ی', regex=False)
                      .str.replace('ك', 'ک', regex=False)
                      .str.replace('\u200c', ' ', regex=False))

df[cols].head(20)
```

```python
for item in cols: 
    print(df[item].value_counts())
    print("------------")
```

```python
print(df[cols].dtypes)
print(df[cols].isnull().sum())
print(df[cols].shape)
print(df[cols].info())
```

```python
print(df[cols].dtypes)
print(df[cols].isnull().sum())
print(df[cols].shape)
print(df[cols].info())
```

```python
# fill by "unknown" for nan value
cols_to_fill = [
    'neighborhood_slug', 'user_type', 'deed_type', 
    'building_direction', 'floor_material'
]

# جایگزینی 'unselect' و NaN با 'unknown'
for col in cols_to_fill:
    df[col] = df[col].replace('unselect', np.nan).fillna('unknown')

print(df[cols].isnull().sum() * 100 / len(df))
```

```python
# delete property_type cause this feature have 97% NULL value 
df = df.drop(columns=['property_type'])
print(df[cols].isnull().sum() * 100 / len(df))
```

```python
nan_rows = df[df['cat3_slug'].isna()]
print(nan_rows)

# delete one row has NAN value in cat3_slug 
df = df.dropna(subset=['cat3_slug'])

# delete few rows has NAN value in city_slug
df = df.dropna(subset=['city_slug'])
```
