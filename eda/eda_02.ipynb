```python
import numpy as np 
import pandas as pandas
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
#load_data 
df = pd.read_parquet('../../data/Divar_cleaned.parquet')
```

```python
fig, ax = plt.subplots(figsize=(14, 6))

# رسم هیستوگرام بر اساس سال ساخت
sns.histplot(data=df, x="construction_year", discrete=True, color="teal", ax=ax)

# تنظیم مرتب‌سازی محور X از کمترین سال به بیشترین
ax.set_xticks(sorted(df["construction_year"].dropna().unique()))
ax.tick_params(axis='x', rotation=45)
sns.despine()


# نمایش برچسب تعداد بالای میله‌ها
ax.bar_label(ax.containers[0], padding=3, rotation=90, fontsize=8)


ax.set_title("Distribution of Building Construction Year (construction_year)")
ax.set_xlabel("Construction Year (Shamsi)")
ax.set_ylabel("Number of Ads")

plt.tight_layout()
plt.show()
```
